1. Process Creation using fork(), exec() and wait()

```python
import os
import sys
import warnings

# Notebook kernels are multi-threaded; silence the harmless fork() warning
warnings.filterwarnings("ignore", category=DeprecationWarning)


def main():
    print(f"Parent process: PID = {os.getpid()}")
    sys.stdout.flush()

    # Pipe lets the parent display the output of the exec'd program,
    # because a notebook does not show output written directly by 'ls'.
    r, w = os.pipe()

    try:
        pid = os.fork()
    except OSError as e:
        print(f"fork failed: {e}")
        return

    if pid == 0:
        # ---- Child process ----
        os.close(r)
        print(f"Child process: PID = {os.getpid()}, Parent PID = {os.getppid()}")
        print("Child is about to exec 'ls -l'")
        sys.stdout.flush()
        os.dup2(w, 1)          # send the program's stdout into the pipe
        os.close(w)
        try:
            os.execlp("ls", "ls", "-l")
        except OSError as e:
            os.write(2, f"exec failed: {e}\n".encode())
            os._exit(1)
    else:
        # ---- Parent process ----
        os.close(w)
        print(f"Parent waiting for child (PID = {pid}) to finish...")
        sys.stdout.flush()
        with os.fdopen(r) as f:
            output = f.read()   # read until the child finishes
        _, status = os.wait()
        print(output, end="")
        if os.WIFEXITED(status):
            print(f"Child exited with status {os.WEXITSTATUS(status)}")
        print("Parent process resumes and exits.")


main()
```

2. Thread Creation using POSIX Threads

```python
import threading

NUM_THREADS = 4


class PartialSum:
    """Argument and result passed to each thread."""

    def __init__(self, start, end):
        self.start = start
        self.end = end
        self.sum = 0


def sum_array(p):
    p.sum = 0
    for i in range(p.start, p.end):
        p.sum += i
    print(f"Thread {threading.get_ident() % 100000}: "
          f"sum[{p.start}..{p.end}) = {p.sum}")


def main():
    n = 1000000
    chunk = n // NUM_THREADS
    data = []
    threads = []

    for i in range(NUM_THREADS):
        start = i * chunk
        end = n if i == NUM_THREADS - 1 else (i + 1) * chunk
        data.append(PartialSum(start, end))
        t = threading.Thread(target=sum_array, args=(data[i],))
        threads.append(t)
        t.start()

    total = 0
    for i in range(NUM_THREADS):
        threads[i].join()
        total += data[i].sum

    print(f"Total sum (0 to {n - 1}) computed using {NUM_THREADS} threads = {total}")


if __name__ == "__main__":
    main()
```

3. CPU Scheduling Algorithms

```python
import copy
from collections import deque

INF = 10 ** 9


class Process:
    def __init__(self, pid, at, bt, pr):
        self.pid, self.at, self.bt, self.pr = pid, at, bt, pr
        self.ct = self.tat = self.wt = self.rem = 0


def print_result(p):
    n = len(p)
    total_wt = total_tat = 0
    print("\nPID\tAT\tBT\tCT\tTAT\tWT")
    for x in p:
        print(f"{x.pid}\t{x.at}\t{x.bt}\t{x.ct}\t{x.tat}\t{x.wt}")
        total_wt += x.wt
        total_tat += x.tat
    print(f"Average Waiting Time    = {total_wt / n:.2f}")
    print(f"Average Turnaround Time = {total_tat / n:.2f}")


def _finish(x, time):
    x.ct = time
    x.tat = x.ct - x.at
    x.wt = x.tat - x.bt


def fcfs(p):
    time = 0
    for x in p:
        if time < x.at:
            time = x.at
        time += x.bt
        _finish(x, time)
    print_result(p)


def _non_preemptive(p, key):
    n = len(p)
    completed, time = 0, 0
    done = [False] * n
    while completed < n:
        idx, best = -1, INF
        for i in range(n):
            if not done[i] and p[i].at <= time and key(p[i]) < best:
                best = key(p[i])
                idx = i
        if idx == -1:
            time += 1
            continue
        time += p[idx].bt
        _finish(p[idx], time)
        done[idx] = True
        completed += 1
    print_result(p)


def sjf(p):
    _non_preemptive(p, lambda x: x.bt)


def priority_scheduling(p):
    _non_preemptive(p, lambda x: x.pr)


def round_robin(p, quantum):
    n = len(p)
    time, completed = 0, 0
    for x in p:
        x.rem = x.bt

    queue = deque()
    in_queue = [False] * n

    # start with process(es) arriving at time 0, in PID order
    for i in range(n):
        if p[i].at <= time:
            queue.append(i)
            in_queue[i] = True

    while completed < n:
        if not queue:
            time += 1
            for i in range(n):
                if not in_queue[i] and p[i].rem > 0 and p[i].at <= time:
                    queue.append(i)
                    in_queue[i] = True
            continue

        idx = queue.popleft()
        slice_ = min(p[idx].rem, quantum)
        time += slice_
        p[idx].rem -= slice_

        for i in range(n):
            if not in_queue[i] and p[i].rem > 0 and p[i].at <= time and i != idx:
                queue.append(i)
                in_queue[i] = True

        if p[idx].rem > 0:
            queue.append(idx)
        else:
            _finish(p[idx], time)
            completed += 1
            in_queue[idx] = False
    print_result(p)


def main():
    n = int(input("Enter number of processes: "))
    p = []
    for i in range(n):
        at, bt, pr = map(int, input(
            f"Process {i + 1} - Arrival Time, Burst Time, Priority: ").replace(",", " ").replace(",", " ").split())
        p.append(Process(i + 1, at, bt, pr))

    print("\nMenu:\n1. FCFS\n2. SJF (Non-preemptive)")
    print("3. Priority (Non-preemptive)\n4. Round Robin")
    choice = int(input("Enter choice: "))

    p = copy.deepcopy(p)
    if choice == 1:
        fcfs(p)
    elif choice == 2:
        sjf(p)
    elif choice == 3:
        priority_scheduling(p)
    elif choice == 4:
        quantum = int(input("Enter time quantum: "))
        round_robin(p, quantum)
    else:
        print("Invalid choice")


if __name__ == "__main__":
    main()
```

4. Producer-Consumer Problem

```python
import threading
import time

BUFFER_SIZE = 5
ITEMS = 10

buffer = [0] * BUFFER_SIZE
in_idx = 0
out_idx = 0

empty = threading.Semaphore(BUFFER_SIZE)
full = threading.Semaphore(0)
mutex = threading.Semaphore(1)


def producer():
    global in_idx
    for i in range(1, ITEMS + 1):
        empty.acquire()
        mutex.acquire()

        buffer[in_idx] = i
        print(f"Producer produced item {i} at slot {in_idx}", flush=True)
        in_idx = (in_idx + 1) % BUFFER_SIZE

        mutex.release()
        full.release()
        time.sleep(0.1)


def consumer():
    global out_idx
    for _ in range(ITEMS):
        full.acquire()
        mutex.acquire()

        item = buffer[out_idx]
        print(f"\t\tConsumer consumed item {item} from slot {out_idx}", flush=True)
        out_idx = (out_idx + 1) % BUFFER_SIZE

        mutex.release()
        empty.release()
        time.sleep(0.15)


def main():
    prod = threading.Thread(target=producer)
    cons = threading.Thread(target=consumer)
    prod.start()
    cons.start()
    prod.join()
    cons.join()
    print("All items produced and consumed.")


if __name__ == "__main__":
    main()
```

5. Readers-Writers Problem

```python
import threading
import time

mutex = threading.Semaphore(1)
wrt = threading.Semaphore(1)
read_count = 0
shared_data = 0


def reader(rid):
    global read_count

    mutex.acquire()
    read_count += 1
    if read_count == 1:
        wrt.acquire()          # first reader locks the resource
    mutex.release()

    print(f"Reader {rid}: read shared_data = {shared_data}", flush=True)
    time.sleep(0.2)

    mutex.acquire()
    read_count -= 1
    if read_count == 0:
        wrt.release()          # last reader unlocks the resource
    mutex.release()


def writer(wid):
    global shared_data

    wrt.acquire()
    shared_data += 1
    print(f"Writer {wid}: wrote shared_data = {shared_data}", flush=True)
    time.sleep(0.3)
    wrt.release()


def main():
    w = [None, None]
    r = []

    w[0] = threading.Thread(target=writer, args=(1,))
    w[0].start()
    for i in range(5):
        t = threading.Thread(target=reader, args=(i + 1,))
        r.append(t)
        t.start()
    w[1] = threading.Thread(target=writer, args=(2,))
    w[1].start()

    for t in r:
        t.join()
    w[0].join()
    w[1].join()


if __name__ == "__main__":
    main()
```

6. Dining Philosophers Problem

```python
import threading
import time

N = 5

forks = [threading.Semaphore(1) for _ in range(N)]
room = threading.Semaphore(N - 1)   # at most N-1 philosophers try to pick up forks


def philosopher(pid):
    left = pid
    right = (pid + 1) % N

    for _ in range(2):
        print(f"Philosopher {pid} is thinking", flush=True)
        time.sleep(0.1)

        room.acquire()            # enter the "room" - avoids deadlock
        forks[left].acquire()     # pick up left fork
        forks[right].acquire()    # pick up right fork

        print(f"Philosopher {pid} is eating (fork {left} and fork {right})", flush=True)
        time.sleep(0.15)

        forks[right].release()
        forks[left].release()
        room.release()

        print(f"Philosopher {pid} finished eating and put down forks", flush=True)


def main():
    philo = [threading.Thread(target=philosopher, args=(i,)) for i in range(N)]
    for t in philo:
        t.start()
    for t in philo:
        t.join()


if __name__ == "__main__":
    main()
```

7. Banker's Algorithm

```python
P = 5   # number of processes
R = 3   # number of resource types

available = [0] * R
maximum = [[0] * R for _ in range(P)]
allocation = [[0] * R for _ in range(P)]
need = [[0] * R for _ in range(P)]


def read_ints(count, prompt=""):
    """Read `count` integers, possibly spread over several input lines."""
    if prompt:
        print(prompt, end="", flush=True)
    vals = []
    while len(vals) < count:
        vals += list(map(int, input().replace(",", " ").replace(",", " ").split()))
    return vals[:count]


def calculate_need():
    for i in range(P):
        for j in range(R):
            need[i][j] = maximum[i][j] - allocation[i][j]


def is_safe():
    """Return the safe sequence (list) if the state is safe, else None."""
    work = available[:]
    finish = [False] * P
    safe_seq = []

    while len(safe_seq) < P:
        found = False
        for i in range(P):
            if not finish[i] and all(need[i][j] <= work[j] for j in range(R)):
                for k in range(R):
                    work[k] += allocation[i][k]
                safe_seq.append(i)
                finish[i] = True
                found = True
        if not found:
            return None   # no process could proceed: unsafe
    return safe_seq


def request_resources(pid, request):
    for j in range(R):
        if request[j] > need[pid][j]:
            print("Error: process has exceeded its maximum claim.")
            return False
        if request[j] > available[j]:
            print("Request denied: not enough resources available.")
            return False

    # tentatively allocate
    for j in range(R):
        available[j] -= request[j]
        allocation[pid][j] += request[j]
        need[pid][j] -= request[j]

    if is_safe() is not None:
        print("Request can be granted safely.")
        return True

    # rollback
    for j in range(R):
        available[j] += request[j]
        allocation[pid][j] -= request[j]
        need[pid][j] += request[j]
    print("Request denied: would lead to an unsafe state.")
    return False


def main():
    global available
    available = read_ints(
        R, f"Enter available resources of each type ({R} types): ")

    print(f"Enter Maximum matrix ({P} processes x {R} resources):")
    for i in range(P):
        maximum[i] = read_ints(R)

    print("Enter Allocation matrix:")
    for i in range(P):
        allocation[i] = read_ints(R)

    calculate_need()

    safe_seq = is_safe()
    if safe_seq is not None:
        print("System is in a SAFE state.")
        print("Safe sequence: " + " ".join(f"P{i}" for i in safe_seq) + " ")
    else:
        print("System is in an UNSAFE state.")

    pid = read_ints(1, "\nEnter process number requesting resources: ")[0]
    request = read_ints(R, f"Enter request vector ({R} values): ")

    request_resources(pid, request)


if __name__ == "__main__":
    main()
```

8. Contiguous Memory Allocation

```python
def _print_allocation(title, psize, allocation):
    print(f"\n--- {title} ---")
    print("Process No.\tProcess Size\tBlock No.")
    for i, size in enumerate(psize):
        if allocation[i] != -1:
            print(f"P{i + 1}\t\t{size}\t\tB{allocation[i] + 1}")
        else:
            print(f"P{i + 1}\t\t{size}\t\tNot Allocated")


def first_fit(bsize, psize):
    block = bsize[:]
    allocation = [-1] * len(psize)
    for i, size in enumerate(psize):
        for j in range(len(block)):
            if block[j] >= size:
                allocation[i] = j
                block[j] -= size
                break
    _print_allocation("First Fit", psize, allocation)


def best_fit(bsize, psize):
    block = bsize[:]
    allocation = [-1] * len(psize)
    for i, size in enumerate(psize):
        best_idx = -1
        for j in range(len(block)):
            if block[j] >= size:
                if best_idx == -1 or block[j] < block[best_idx]:
                    best_idx = j
        if best_idx != -1:
            allocation[i] = best_idx
            block[best_idx] -= size
    _print_allocation("Best Fit", psize, allocation)


def worst_fit(bsize, psize):
    block = bsize[:]
    allocation = [-1] * len(psize)
    for i, size in enumerate(psize):
        worst_idx = -1
        for j in range(len(block)):
            if block[j] >= size:
                if worst_idx == -1 or block[j] > block[worst_idx]:
                    worst_idx = j
        if worst_idx != -1:
            allocation[i] = worst_idx
            block[worst_idx] -= size
    _print_allocation("Worst Fit", psize, allocation)


def main():
    bn = int(input("Enter number of memory blocks: "))
    bsize = list(map(int, input("Enter sizes of blocks: ").replace(",", " ").replace(",", " ").split()))[:bn]

    pn = int(input("Enter number of processes: "))
    psize = list(map(int, input("Enter sizes of processes: ").replace(",", " ").replace(",", " ").split()))[:pn]

    first_fit(bsize, psize)
    best_fit(bsize, psize)
    worst_fit(bsize, psize)


if __name__ == "__main__":
    main()
```

9. Page Replacement Algorithms

```python
def fifo(ref, nf):
    frames = [-1] * nf
    front, faults = 0, 0
    for page in ref:
        if page not in frames:
            frames[front] = page
            front = (front + 1) % nf
            faults += 1
    print(f"FIFO Page Faults  = {faults}")


def lru(ref, nf):
    frames = [-1] * nf
    last_used = [0] * nf
    filled, faults = 0, 0

    for i, page in enumerate(ref):
        pos = frames.index(page) if page in frames else -1

        if pos == -1:
            faults += 1
            if filled < nf:
                frames[filled] = page
                last_used[filled] = i
                filled += 1
            else:
                # find least recently used frame
                lru_idx = 0
                for j in range(1, nf):
                    if last_used[j] < last_used[lru_idx]:
                        lru_idx = j
                frames[lru_idx] = page
                last_used[lru_idx] = i
        else:
            last_used[pos] = i
    print(f"LRU Page Faults   = {faults}")


def optimal(ref, nf):
    n = len(ref)
    frames = [-1] * nf
    filled, faults = 0, 0

    for i in range(n):
        if ref[i] in frames[:filled]:
            continue

        faults += 1
        if filled < nf:
            frames[filled] = ref[i]
            filled += 1
            continue

        # replace the page used farthest in the future (or never again)
        replace_idx, farthest = -1, -1
        for j in range(nf):
            k = i + 1
            while k < n and frames[j] != ref[k]:
                k += 1
            if k == n:                 # never used again
                replace_idx = j
                break
            if k > farthest:
                farthest = k
                replace_idx = j
        frames[replace_idx] = ref[i]
    print(f"Optimal Page Faults = {faults}")


def main():
    n = int(input("Enter number of pages in reference string: "))
    ref = list(map(int, input("Enter the reference string: ").replace(",", " ").replace(",", " ").split()))[:n]
    nf = int(input("Enter number of frames: "))

    fifo(ref, nf)
    lru(ref, nf)
    optimal(ref, nf)


if __name__ == "__main__":
    main()
```

10. Disk Scheduling Algorithms

```python
def fcfs(req, head):
    total, cur = 0, head
    for r in req:
        total += abs(cur - r)
        cur = r
    print(f"FCFS   total head movement = {total}")


def sstf(req, head):
    n = len(req)
    visited = [False] * n
    total, cur = 0, head
    for _ in range(n):
        idx, best = -1, 1 << 30
        for i in range(n):
            if not visited[i] and abs(cur - req[i]) < best:
                best = abs(cur - req[i])
                idx = i
        visited[idx] = True
        total += best
        cur = req[idx]
    print(f"SSTF   total head movement = {total}")


def scan(req, head, disk_size):
    arr = sorted(req + [head])
    idx = arr.index(head)

    total, cur = 0, head
    # go towards the higher end first
    for i in range(idx + 1, len(arr)):
        total += abs(cur - arr[i])
        cur = arr[i]
    total += abs(cur - (disk_size - 1))
    cur = disk_size - 1
    for i in range(idx - 1, -1, -1):
        total += abs(cur - arr[i])
        cur = arr[i]
    print(f"SCAN   total head movement = {total}")


def cscan(req, head, disk_size):
    arr = sorted(req + [head])
    idx = arr.index(head)

    total, cur = 0, head
    for i in range(idx + 1, len(arr)):
        total += abs(cur - arr[i])
        cur = arr[i]
    total += abs(cur - (disk_size - 1))   # go to the end
    total += disk_size - 1                # jump to 0 (cylinder 0)
    cur = 0
    for i in range(idx):
        total += abs(cur - arr[i])
        cur = arr[i]
    print(f"C-SCAN total head movement = {total}")


def main():
    n = int(input("Enter number of requests: "))
    req = list(map(int, input("Enter the requests: ").replace(",", " ").replace(",", " ").split()))[:n]
    head = int(input("Enter initial head position: "))
    disk_size = int(input("Enter disk size (max cylinder + 1): "))

    fcfs(req, head)
    sstf(req, head)
    scan(req, head, disk_size)
    cscan(req, head, disk_size)


if __name__ == "__main__":
    main()
```

11. File Allocation Methods

```python
MAXBLOCKS = 30

disk = [0] * MAXBLOCKS   # 0 = free, 1 = allocated


def reset_disk():
    for i in range(MAXBLOCKS):
        disk[i] = 0


def contiguous_allocation():
    reset_disk()
    print("\n--- Contiguous Allocation ---")
    nf = int(input("Enter number of files: "))

    for f in range(nf):
        start, length = map(int, input(
            f"File {f + 1} - starting block, length: ").replace(",", " ").replace(",", " ").split())

        ok = True
        for i in range(start, start + length):
            if i < 0 or i >= MAXBLOCKS or disk[i] == 1:
                ok = False
                break

        if not ok:
            print("Allocation failed: blocks not free/out of range.")
            continue
        for i in range(start, start + length):
            disk[i] = 1
        print(f"File {f + 1} allocated blocks {start} to {start + length - 1}")


def linked_allocation():
    reset_disk()
    print("\n--- Linked Allocation ---")
    nf = int(input("Enter number of files: "))

    for f in range(nf):
        length = int(input(f"File {f + 1} - number of blocks: "))

        out = f"File {f + 1} allocated blocks: "
        prev, count = -1, 0
        i = 0
        while i < MAXBLOCKS and count < length:
            if disk[i] == 0:
                disk[i] = 1
                out += f"{i} "
                if prev != -1:
                    out += "-> "
                prev = i
                count += 1
            i += 1
        out += "(NULL)"
        print(out)
        if count < length:
            print("Warning: disk full, could not allocate all blocks.")


def indexed_allocation():
    reset_disk()
    print("\n--- Indexed Allocation ---")
    nf = int(input("Enter number of files: "))

    for f in range(nf):
        length = int(input(f"File {f + 1} - number of data blocks needed: "))

        index_block = -1
        for i in range(MAXBLOCKS):
            if disk[i] == 0:
                index_block = i
                disk[i] = 1
                break

        out = f"File {f + 1} - Index block: {index_block}, Data blocks: "
        count = 0
        i = 0
        while i < MAXBLOCKS and count < length:
            if disk[i] == 0:
                disk[i] = 1
                out += f"{i} "
                count += 1
            i += 1
        print(out)


def main():
    print("Menu:\n1. Contiguous Allocation\n2. Linked Allocation")
    print("3. Indexed Allocation")
    choice = int(input("Enter choice: "))

    if choice == 1:
        contiguous_allocation()
    elif choice == 2:
        linked_allocation()
    elif choice == 3:
        indexed_allocation()
    else:
        print("Invalid choice")


if __name__ == "__main__":
    main()
```
